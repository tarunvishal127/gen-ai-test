Exercise - 1: Basic Data Structures
To create lists, use aggregate functions, arithmetic, logical operators, contol flows and functions.

```python
num = [-10,-3,0,1,2,7,9]

cities = {
    "Hyd":8000000,
    "Blr": 10000000,
    "Chn": 9000000,
    "Mum": 12000000,
    "Del": 11000000
}

def sum_and_avg(nums: list[int]):
    total = sum(nums)
    average = total / len(nums) if nums else 0
    return total, average
def city_with_max_population(cities: dict[str, int]):
    max_city = max(cities, key=cities.get) if cities else None
    return max_city
def large_cities(cities: dict[str, int], threshold: int = 10000000):
    large_cities_list = [city for city, population in cities.items() if population > threshold]
    return large_cities_list

total, average = sum_and_avg(num)
print(f"Total: {total}, Average: {average}")

max_city = city_with_max_population(cities)
print(f"City with maximum population: {max_city}")

large_cities_list = large_cities(cities)
print(f"Large cities: {large_cities_list}")
```

Exercise - 2:
Control Flow and Functions

```python
def classify_number(num: int):
    if num < 0:
        return "Negative"
    elif num == 0:
        return "Zero"
    else:
        return "Positive"

nums = [-10, -3, 0, 1, 2, 7, 9]
counter = {"Negative": 0, "Zero": 0, "Positive": 0}
for n in nums:
    classification = classify_number(n)
    counter[classification] += 1
print(counter)
```

Exercise - 3:
Comprehensions

```python
labels = [classify_number(n) for n in nums]
counts = {
    "negative": sum(1 for l in labels if l == "Negative"),
    "zero": sum(1 for l in labels if l == "Zero"),
    "positive": sum(1 for l in labels if l == "Positive"),
}
print(counts)

nums1 = list(range(-10, 11))
squares = [n**2 for n in nums1]
print(squares)
even_squares = [n**2 for n in nums1 if n % 2 == 0]
print(even_squares)
odd_squares = [n**2 for n in nums1 if n % 2 != 0]
print(odd_squares)
squares_dict = {n: n**2 for n in nums1}
print(squares_dict)
squares_div5 = {n: n**2 for n in nums1 if n % 5 == 0}
print(squares_div5)

print(len(squares))
```

Exercise - 4:
File I/O

```python
def analyze_text_file(path: str):
    with open(path, 'r') as file:
        text = file.readlines()
        num_lines = len(text)
        num_words = sum(len(line.split()) for line in text)
        num_chars = sum(len(line) for line in text)
    return num_lines, num_words, num_chars
stats = analyze_text_file('/Users/tarun/tarun/genai/code/gen-ai-test/week01/scripts/sample_data.txt')
print(f"Lines: {stats[0]}, Words: {stats[1]}, Characters: {stats[2]}")

def save_stats(stats:dict, out_path:str):
    with open(out_path, 'w') as file:
            file.write(f"Lines: {stats[0]}\nWords: {stats[1]}\nCharacters: {stats[2]}\n")

save_stats(stats, '/Users/tarun/tarun/genai/code/gen-ai-test/week01/scripts/day_01_stats.txt')
```

Exercise - 5:
Small Data Task

```python
records = [
    {"name": "Alice", "age": 29, "city": "Delhi", "salary": 70000},
    {"name": "Bob", "age": 35, "city": "Mumbai", "salary": 90000},
    {"name": "Chitra", "age": 24, "city": "Delhi", "salary": 50000},
    {"name": "Dev", "age": 40, "city": "Bangalore", "salary": 120000},
    {"name": "Esha", "age": 27, "city": "Mumbai", "salary": 65000},
    {"name": "Farhan", "age": 33, "city": "Hyderabad", "salary": 80000},
]

def average_salary(records: list[dict]):
    total_salary = sum(record["salary"] for record in records)
    average = total_salary / len(records) if records else 0
    return average
def city_with_highest_avg_salary(records: list[dict]):
    city_salary = {}
    city_count = {}
    for record in records:
        city = record["city"]
        salary = record["salary"]
        city_salary[city] = city_salary.get(city, 0) + salary
        city_count[city] = city_count.get(city, 0) + 1
    avg_salary_by_city = {city: city_salary[city] / city_count[city] for city in city_salary}
    max_city = max(avg_salary_by_city, key=avg_salary_by_city.get) if avg_salary_by_city else None
    return max_city
def filter_records_by_age(records: list[dict], age_threshold: int):
    filtered_records = [record for record in records if record["age"] > age_threshold]
    return filtered_records

avg_salary = average_salary(records)
print(f"Average Salary: {avg_salary}")
highest_avg_salary_city = city_with_highest_avg_salary(records)
print(f"City with highest average salary: {highest_avg_salary_city}")
filtered_records = filter_records_by_age(records, 30)
print(f"Records with age > 30: {len(filtered_records)}")
```
